```python
import json
import os
import sys

os.chdir(r"D:\A_scientific_learning\weather-outfit-advisor")
if os.getcwd() not in sys.path:
    sys.path.insert(0, os.getcwd())

from src.agent.outfit_agent import get_outfit_advice

with open("data/validation_cases.json", "r", encoding="utf-8") as f:
    cases = json.load(f)

results = []
for case in cases:
    print(f"\n{'='*60}")
    print(f"用例 {case['id']}: {case['city']} / {case['scene']} / {case['climate']}")
    print('='*60)
    try:
        advice = get_outfit_advice(case["city"], case["date"], case["scene"])
        results.append({"case": case, "advice": advice, "error": None})
        print(advice)
    except Exception as e:
        results.append({"case": case, "advice": None, "error": str(e)})
        print(f"报错: {e}")

print(f"\n\n{'='*60}")
print(f"完成：{sum(1 for r in results if r['error'] is None)}/{len(results)} 成功")
print('='*60)
```

```python
import json
import os

outputs = []
for case in cases:
    try:
        advice = get_outfit_advice(case["city"], case["date"], case["scene"])
        outputs.append({
            "case_id": case["id"],
            "city": case["city"],
            "scene": case["scene"],
            "climate": case["climate"],
            "advice": advice,
            "error": None
        })
    except Exception as e:
        outputs.append({
            "case_id": case["id"],
            "city": case["city"],
            "scene": case["scene"],
            "climate": case["climate"],
            "advice": None,
            "error": str(e)
        })

with open("data/validation_outputs.json", "w", encoding="utf-8") as f:
    json.dump(outputs, f, ensure_ascii=False, indent=2)

print(f"已保存 {len(outputs)} 条输出到 data/validation_outputs.json")
```
